```python
duplicate_columns = []

cols = df.columns

for i in range(len(cols)):
    for j in range(i + 1, len(cols)):
        if df[cols[i]].equals(df[cols[j]]):
            duplicate_columns.append(
                (cols[i], cols[j]))

duplicate_columns
```

## Data Cleaning


**Remove completely empty columns**

```python
df = df.dropna(axis=1, how="all")
```

```python
# drop the duplicate columns
df = df.drop(columns=[
    "PipelineInfo_ClientCountry.1",
    "Tenant_Country.1"
], errors="ignore")
```

```python
#Remove CPU Architecture because it is almost completely missing
df = df.drop(columns=["R7_DeviceInfo_CpuArchitecture"], errors="ignore")
```

```python
df["Feedback_DateTime"] = pd.to_datetime(
    df["Feedback_DateTime"],
    errors="coerce"
)
```

```python
df["FirstSeenTime_Date"] = pd.to_datetime(
    df["FirstSeenTime_Date"],
    errors="coerce"
)
```

```python
df.loc[
    df["FirstSeenTime_Date"] == "1970-01-01",
    "FirstSeenTime_Date"
] = pd.NaT
```

```python
df = df.drop_duplicates()
```

```python
print("Final shape:", df.shape)
print("Duplicate rows:", df.duplicated().sum())
print("Missing Feedback IDs:", df["Feedback_ID"].isna().sum())
print("Missing Ratings:", df["Feedback_Rating"].isna().sum())
print("Missing ResponseType:", df["ResponseType"].isna().sum())
```
