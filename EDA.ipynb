```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.max_columns", 100)
pd.set_option("display.max_rows", 100)

raw = pd.read_csv("Dataset_clean.csv")

analysis = raw.copy()
analysis["Feedback_DateTime"] = pd.to_datetime(analysis["Feedback_DateTime"], errors="coerce")
analysis["FirstSeenTime_Date"] = pd.to_datetime(analysis["FirstSeenTime_Date"], errors="coerce")
analysis["Feedback_Month_Name"] = analysis["Feedback_DateTime"].dt.strftime("%b")
analysis["Feedback_Day"] = analysis["Feedback_DateTime"].dt.date

categorical_columns = [
    "PipelineInfo_ClientCountry", "Tenant_Country",
    "Tenant_MSSalesVerticalName", "Tenant_CustomerSegmentGroup",
    "NewUser_Period"
]
for column in categorical_columns:
    analysis[column] = analysis[column].fillna("Unknown").astype(str).str.strip()

analysis["Derived_ResponseType"] = pd.cut(
    analysis["Feedback_Rating"],
    bins=[-1, 6, 8, 10],
    labels=["Detractor", "Passive", "Promoter"]
).astype("string")

analysis["Response_Match"] = analysis["ResponseType"].eq(analysis["Derived_ResponseType"])
print(f"Cleaned shape: {analysis.shape}")
print(f"Response label mismatches: {(~analysis['Response_Match']).sum()}")
```

```python
def nps_score(frame):
    promoters = (frame["Feedback_Rating"] >= 9).sum()
    detractors = (frame["Feedback_Rating"] <= 6).sum()
    total = frame["Feedback_Rating"].notna().sum()
    return 100 * (promoters - detractors) / total if total else np.nan

baseline = pd.DataFrame({
    "Metric": ["Responses", "Promoters", "Passives", "Detractors", "NPS"],
    "Value": [
        len(analysis),
        (analysis["ResponseType"] == "Promoter").sum(),
        (analysis["ResponseType"] == "Passive").sum(),
        (analysis["ResponseType"] == "Detractor").sum(),
        round(nps_score(analysis), 2),
    ],
})
baseline["Share_%"] = np.nan
baseline.loc[1:3, "Share_%"] = (baseline.loc[1:3, "Value"] / len(analysis) * 100).round(2)
print("NPS baseline")
display(baseline)

monthly_counts = (
    analysis.groupby("Feedback_Month_Name", sort=False)
    .agg(
        Responses=("Feedback_ID", "size"),
        Promoters=("Feedback_Rating", lambda values: (values >= 9).sum()),
        Detractors=("Feedback_Rating", lambda values: (values <= 6).sum()),
    )
    .reindex(["Dec", "Jan", "Feb"])
)
monthly_counts["NPS"] = (
    100 * (monthly_counts["Promoters"] - monthly_counts["Detractors"])
    / monthly_counts["Responses"]
).round(2)
monthly = monthly_counts[["Responses", "NPS"]]
print("Monthly trend")
display(monthly)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
monthly["NPS"].plot(kind="bar", ax=axes[0], color="#1677ff", title="NPS by feedback month")
axes[0].axhline(0, color="black", linewidth=0.8)
axes[0].set_ylabel("NPS")
axes[0].set_xlabel("")
axes[0].tick_params(axis="x", rotation=0)
analysis["ResponseType"].value_counts().reindex(["Detractor", "Passive", "Promoter"]).plot(
    kind="bar", ax=axes[1], color=["#d1495b", "#f4a261", "#2a9d8f"], title="Response mix"
)
axes[1].set_ylabel("Responses")
axes[1].set_xlabel("")
axes[1].tick_params(axis="x", rotation=0)
plt.tight_layout()
plt.show()
```

## Who gives each response type?

The following tables describe the people and customer contexts behind Promoter, Passive, and Detractor responses. Segment shares are calculated within each response type, and small segments are excluded from the ranked view.

```python
response_order = ["Promoter", "Passive", "Detractor"]

response_profile = (
    analysis.groupby("ResponseType", observed=False)
    .agg(
        Responses=("Feedback_ID", "size"),
        Avg_Rating=("Feedback_Rating", "mean"),
        Avg_ActiveEvents=("ActiveEvents", "mean"),
        Avg_CollabEvents=("CollabEvents", "mean"),
        Avg_ReadEvents=("ReadEvents", "mean"),
        Avg_ActiveChannels=("ActiveChannels", "mean"),
        Avg_ActiveTeams=("ActiveTeams", "mean"),
    )
    .reindex(response_order)
)
response_profile["Share_%"] = (
    response_profile["Responses"] / response_profile["Responses"].sum() * 100
)
response_profile = response_profile[
    ["Responses", "Share_%", "Avg_Rating", "Avg_ActiveEvents", "Avg_CollabEvents",
     "Avg_ReadEvents", "Avg_ActiveChannels", "Avg_ActiveTeams"]
].round(2)

print("Response-type profile")
display(response_profile)

segment_columns = [
    "NewUser_Period",
    "Tenant_CustomerSegmentGroup",
    "Tenant_Country",
    "Tenant_MSSalesVerticalName",
]
segment_rows = []
for segment_column in segment_columns:
    counts = pd.crosstab(analysis[segment_column], analysis["ResponseType"])
    counts["Responses"] = counts.sum(axis=1)
    eligible = counts[counts["Responses"] >= 100]
    for response_type in response_order:
        if response_type not in eligible.columns:
            continue
        top_segments = eligible.sort_values(response_type, ascending=False).head(5)
        response_total = response_profile.loc[response_type, "Responses"]
        for segment, row in top_segments.iterrows():
            segment_rows.append({
                "Dimension": segment_column,
                "ResponseType": response_type,
                "Segment": segment,
                "Responses": int(row[response_type]),
                "Share_of_response_type_%": round(row[response_type] / response_total * 100, 2),
                "Segment_total": int(row["Responses"]),
            })

segment_profile = pd.DataFrame(segment_rows)
print("Largest sufficiently sized segments within each response type")
display(segment_profile)
```

```python
behavior_metrics = [
    "ActiveEvents", "CollabEvents", "ReadEvents", "MobileEvents", "DesktopEvents",
    "WebEvents", "ChatCreateEvents", "ChannelCreateEvents", "FileCreateEvents",
    "CMDCreateEvents", "PlatformCreateEvents", "ActiveChannels", "ActiveTeams"
]
behavior_comparison = analysis.groupby("ResponseType")[behavior_metrics].agg(["mean", "median"])
comparison_table = behavior_comparison.T.loc[:, ["Promoter", "Detractor"]]
comparison_table["Difference"] = comparison_table["Promoter"] - comparison_table["Detractor"]
print("Behavior comparison: promoter versus detractor")
display(comparison_table.sort_values("Difference"))

behavior_corr = (
    analysis[behavior_metrics + ["Feedback_Rating"]]
    .corr(method="spearman")["Feedback_Rating"]
    .drop("Feedback_Rating")
    .sort_values()
    .to_frame("Spearman correlation with rating")
)
print("Behavior associations with rating")
display(behavior_corr)

log_behavior = np.log1p(analysis[behavior_metrics])
log_behavior["ResponseType"] = analysis["ResponseType"].to_numpy()
log_behavior.boxplot(column="ActiveEvents", by="ResponseType", figsize=(7, 4))
plt.suptitle("")
plt.title("Active events by response type (log scale)")
plt.xlabel("")
plt.ylabel("log(1 + ActiveEvents)")
plt.show()
```

## Do usage patterns hold across customer segments?

Compare Promoter and Detractor behavior within each sufficiently sized customer segment. Positive values mean Detractors show higher usage than Promoters.

```python
segment_behavior_metrics = [
    "ActiveEvents", "DesktopEvents", "ReadEvents", "CollabEvents",
    "ChatCreateEvents", "ActiveChannels", "ActiveTeams"
]
segment_behavior_rows = []

for segment in ["Tenant_CustomerSegmentGroup", "NewUser_Period"]:
    segment_counts = analysis[segment].value_counts()
    eligible_segments = segment_counts[segment_counts >= 150].index
    for segment_value in eligible_segments:
        segment_frame = analysis[analysis[segment] == segment_value]
        grouped = segment_frame.groupby("ResponseType")[segment_behavior_metrics].mean()
        if not {"Promoter", "Detractor"}.issubset(grouped.index):
            continue
        promoter_count = (segment_frame["ResponseType"] == "Promoter").sum()
        detractor_count = (segment_frame["ResponseType"] == "Detractor").sum()
        row = {
            "Dimension": segment,
            "Segment": segment_value,
            "Responses": len(segment_frame),
            "Segment_NPS": round(100 * (promoter_count - detractor_count) / len(segment_frame), 2),
            "Promoter_responses": promoter_count,
            "Detractor_responses": detractor_count,
        }
        for metric in segment_behavior_metrics:
            row[f"Detractor_minus_Promoter_{metric}"] = round(
                grouped.loc["Detractor", metric] - grouped.loc["Promoter", metric], 2
            )
        segment_behavior_rows.append(row)

segment_behavior_comparison = pd.DataFrame(segment_behavior_rows)
print("Within-segment usage differences: Detractor minus Promoter")
display(segment_behavior_comparison)

pattern_consistency = (
    segment_behavior_comparison
    .groupby("Dimension")[[f"Detractor_minus_Promoter_{metric}" for metric in segment_behavior_metrics]]
    .agg(["mean", "min", "max"])
    .round(2)
)
print("Pattern consistency across eligible segments")
display(pattern_consistency)
```

## Strongest observable usage differences

Rank the usage measures by size and repeatability. These are descriptive associations, not causal effects.

```python
strongest_rows = []
for metric in behavior_metrics:
    promoter_values = analysis.loc[analysis["ResponseType"] == "Promoter", metric]
    detractor_values = analysis.loc[analysis["ResponseType"] == "Detractor", metric]
    promoter_mean = promoter_values.mean()
    detractor_mean = detractor_values.mean()
    pooled_sd = np.sqrt((promoter_values.var() + detractor_values.var()) / 2)
    segment_column = f"Detractor_minus_Promoter_{metric}"
    if segment_column in segment_behavior_comparison.columns:
        customer_segment_values = segment_behavior_comparison.loc[
            segment_behavior_comparison["Dimension"] == "Tenant_CustomerSegmentGroup",
            segment_column,
        ]
        segments_detractor_higher = int((customer_segment_values > 0).sum())
        segments_compared = int(customer_segment_values.notna().sum())
    else:
        segments_detractor_higher = np.nan
        segments_compared = np.nan
    strongest_rows.append({
        "Metric": metric,
        "Promoter_mean": promoter_mean,
        "Detractor_mean": detractor_mean,
        "Absolute_gap": detractor_mean - promoter_mean,
        "Gap_vs_Promoter_%": 100 * (detractor_mean - promoter_mean) / promoter_mean if promoter_mean else np.nan,
        "Spearman_r": behavior_corr.loc[metric, "Spearman correlation with rating"],
        "Segments_Detractor_higher": segments_detractor_higher,
        "Segments_compared": segments_compared,
        "Standardized_gap": (detractor_mean - promoter_mean) / pooled_sd if pooled_sd else np.nan,
    })

strongest_differences = (
    pd.DataFrame(strongest_rows)
    .sort_values("Standardized_gap", ascending=False)
    .round(3)
)
print("Strongest observable differences, ranked by standardized gap")
display(strongest_differences)

print("Largest repeatable volume differences")
display(
    strongest_differences[
        ["Metric", "Absolute_gap", "Gap_vs_Promoter_%", "Segments_Detractor_higher", "Segments_compared"]
    ].sort_values("Absolute_gap", ascending=False).head(7)
)
```

## Potential Teams optimization areas

These recommendations translate observable usage differences into hypotheses for investigation. The current dataset does not identify the specific feature or defect causing dissatisfaction.

| Observable signal | Potential optimization area | What to investigate next |
|---|---|---|
| Detractors have about 10.6% more total activity and 10.6% more desktop activity than Promoters, consistently across major segments. | **Desktop performance and reliability for high-volume users** | Connect NPS to startup time, latency, crashes, CPU or memory pressure, sync failures, and error telemetry. Prioritize workflows with repeated or long sessions. |
| Detractors have 13.7% more read events. | **Information retrieval and notification workflows** | Check search latency, notification overload, unread-count behavior, navigation time, and whether users repeatedly reopen or scan content without completing tasks. |
| Detractors have 14.0% more collaboration events and 15.4% more chat creation events. | **High-frequency collaboration and messaging flows** | Examine message send failures, delivery delay, compose friction, thread navigation, mentions, and attachment or sharing reliability. |
| Detractors show higher usage volume but only weak correlations with rating, all below 0.06 in magnitude. | **Friction reduction rather than engagement maximization** | Avoid optimizing for more events. Test whether reducing time, errors, retries, or effort in high-volume workflows improves NPS and task success. |
| EDU is the exception for active channels and teams, where Detractors are slightly lower than Promoters. | **Segment-specific breadth and onboarding review** | Investigate whether EDU users have different collaboration structures, permissions, or adoption patterns before applying enterprise-wide fixes. |

Recommended validation: link these cohorts to product-quality telemetry and customer verbatims, then run a targeted experiment or matched pre/post analysis. Track NPS, detractor rate, task completion, latency, reliability, and support contacts by customer segment.

```python
overall_nps = nps_score(analysis)

def segment_summary(frame, dimension, minimum_responses=150):
    summary = frame.groupby(dimension).agg(
        Responses=("Feedback_ID", "size"),
        Promoters=("Feedback_Rating", lambda values: (values >= 9).sum()),
        Detractors=("Feedback_Rating", lambda values: (values <= 6).sum()),
        Avg_Rating=("Feedback_Rating", "mean"),
    )
    summary["NPS"] = 100 * (summary["Promoters"] - summary["Detractors"]) / summary["Responses"]
    summary["Detractor_Rate_%"] = 100 * summary["Detractors"] / summary["Responses"]
    summary["Gap_vs_Overall"] = summary["NPS"] - overall_nps
    summary["Opportunity_Index"] = (
        (overall_nps - summary["NPS"]).clip(lower=0)
        * summary["Responses"] / len(frame)
    )
    return (
        summary[summary["Responses"] >= minimum_responses]
        .sort_values("Opportunity_Index", ascending=False)
        .round({"NPS": 2, "Detractor_Rate_%": 2, "Gap_vs_Overall": 2, "Opportunity_Index": 2, "Avg_Rating": 2})
    )

segment_dimensions = {
    "Tenant country": "Tenant_Country",
    "Client country": "PipelineInfo_ClientCountry",
    "Customer segment": "Tenant_CustomerSegmentGroup",
    "Industry": "Tenant_MSSalesVerticalName",
    "User period": "NewUser_Period",
}
segment_tables = {}
for label, dimension in segment_dimensions.items():
    segment_tables[label] = segment_summary(analysis, dimension)
    print(f"{label}: highest opportunity segments (minimum 150 responses)")
    display(segment_tables[label].head(10))

country_plot = segment_tables["Tenant country"].head(10).sort_values("NPS")
country_plot["NPS"].plot(kind="barh", figsize=(8, 5), color="#d1495b")
plt.axvline(overall_nps, color="#1677ff", linestyle="--", label=f"Overall NPS {overall_nps:.1f}")
plt.title("Largest tenant-country opportunities")
plt.xlabel("NPS")
plt.legend()
plt.tight_layout()
plt.show()
```

## Opportunity ranking: gap, reach, and differentiation

A priority score combines the positive NPS gap versus the overall baseline, the segment's share of responses, and the strength of its within-segment Promoter/Detractor behavior difference. This is a prioritization heuristic, not a causal impact estimate.

```python
ranking_metrics = [
    "ActiveEvents", "DesktopEvents", "ReadEvents", "CollabEvents", "ChatCreateEvents"
]
ranking_dimensions = {
    "Tenant country": "Tenant_Country",
    "Customer segment": "Tenant_CustomerSegmentGroup",
    "Industry": "Tenant_MSSalesVerticalName",
}
ranking_rows = []

for label, dimension in ranking_dimensions.items():
    for segment_value, segment_frame in analysis.groupby(dimension, dropna=False):
        if len(segment_frame) < 150:
            continue
        promoters = segment_frame[segment_frame["ResponseType"] == "Promoter"]
        detractors = segment_frame[segment_frame["ResponseType"] == "Detractor"]
        if promoters.empty or detractors.empty:
            continue
        segment_nps = 100 * (len(promoters) - len(detractors)) / len(segment_frame)
        nps_gap = max(overall_nps - segment_nps, 0)
        differentiation_values = []
        for metric in ranking_metrics:
            promoter_values = promoters[metric]
            detractor_values = detractors[metric]
            pooled_sd = np.sqrt((promoter_values.var() + detractor_values.var()) / 2)
            if pooled_sd > 0:
                differentiation_values.append(
                    abs(detractor_values.mean() - promoter_values.mean()) / pooled_sd
                )
        differentiation_strength = np.mean(differentiation_values) if differentiation_values else np.nan
        reach_fraction = len(segment_frame) / len(analysis)
        ranking_rows.append({
            "Dimension": label,
            "Segment": segment_value,
            "Responses": len(segment_frame),
            "Reach_%": 100 * reach_fraction,
            "Segment_NPS": segment_nps,
            "NPS_gap_vs_overall": nps_gap,
            "Differentiation_strength": differentiation_strength,
            "Priority_score": nps_gap * reach_fraction * differentiation_strength,
        })

opportunity_ranking = (
    pd.DataFrame(ranking_rows)
    .sort_values("Priority_score", ascending=False)
    .round({
        "Reach_%": 2,
        "Segment_NPS": 2,
        "NPS_gap_vs_overall": 2,
        "Differentiation_strength": 3,
        "Priority_score": 3,
    })
)
print("Ranked opportunities using NPS gap, reach, and differentiation")
display(opportunity_ranking.head(15))
```

## Context-adjusted Promoter versus Detractor baseline

This exploratory comparison controls for customer context by comparing Promoters and Detractors within country, customer segment, industry, user period, and month strata. It excludes `Feedback_Rating` and `ResponseType` from usage measures and excludes Passives from this binary comparison. It is descriptive, not causal.

```python
adjustment_dimensions = [
    "Tenant_Country", "Tenant_CustomerSegmentGroup",
    "Tenant_MSSalesVerticalName", "NewUser_Period", "Feedback_Month_Name",
]
adjustment_metrics = [
    "ActiveEvents", "DesktopEvents", "ReadEvents", "CollabEvents", "ChatCreateEvents"
]
context_frame = analysis[analysis["ResponseType"].isin(["Promoter", "Detractor"])].copy()
context_frame["Context_Key"] = context_frame[adjustment_dimensions].astype(str).agg("|".join, axis=1)

adjusted_rows = []
for metric in adjustment_metrics:
    context_gaps = []
    for _, context_group in context_frame.groupby("Context_Key"):
        promoters = context_group.loc[context_group["ResponseType"] == "Promoter", metric]
        detractors = context_group.loc[context_group["ResponseType"] == "Detractor", metric]
        if len(promoters) < 5 or len(detractors) < 5:
            continue
        pooled_sd = np.sqrt((promoters.var() + detractors.var()) / 2)
        if pooled_sd == 0:
            continue
        context_gaps.append({
            "gap": detractors.mean() - promoters.mean(),
            "standardized_gap": (detractors.mean() - promoters.mean()) / pooled_sd,
            "weight": len(context_group),
        })
    context_gaps = pd.DataFrame(context_gaps)
    adjusted_rows.append({
        "Metric": metric,
        "Contexts_compared": len(context_gaps),
        "Adjusted_gap": np.average(context_gaps["gap"], weights=context_gaps["weight"])
        if not context_gaps.empty else np.nan,
        "Adjusted_standardized_gap": np.average(
            context_gaps["standardized_gap"], weights=context_gaps["weight"]
        ) if not context_gaps.empty else np.nan,
    })

adjusted_usage_associations = (
    pd.DataFrame(adjusted_rows)
    .sort_values("Adjusted_standardized_gap", ascending=False)
    .round(3)
)
print("Context-adjusted usage differences: Detractor minus Promoter")
display(adjusted_usage_associations)
```

```python
def add_nps_confidence_interval(summary, frame, dimension, z=1.96):
    rates = frame.groupby(dimension)["Feedback_Rating"].agg(
        n="size",
        promoter_rate=lambda values: (values >= 9).mean(),
        detractor_rate=lambda values: (values <= 6).mean(),
    )
    rates["nps"] = 100 * (rates["promoter_rate"] - rates["detractor_rate"])
    rates["nps_se"] = 100 * np.sqrt(
        (rates["promoter_rate"] + rates["detractor_rate"] -
         (rates["promoter_rate"] - rates["detractor_rate"]) ** 2) / rates["n"]
    )
    rates["NPS_CI_Lower"] = rates["nps"] - z * rates["nps_se"]
    rates["NPS_CI_Upper"] = rates["nps"] + z * rates["nps_se"]
    return summary.join(rates[["NPS_CI_Lower", "NPS_CI_Upper"]]).round(2)

country_with_ci = add_nps_confidence_interval(
    segment_tables["Tenant country"], analysis, "Tenant_Country"
)
print("Tenant-country opportunities with approximate 95% NPS confidence intervals")
display(country_with_ci.head(10))

print("Segments with confidence intervals entirely below overall NPS")
stable_country_opportunities = country_with_ci[
    country_with_ci["NPS_CI_Upper"] < overall_nps
].sort_values("Opportunity_Index", ascending=False)
display(stable_country_opportunities.head(10))
```

## Findings and recommended next steps

### What differentiates promoters and detractors

- Detractors are more active in the preceding seven-day window than promoters: mean `ActiveEvents` is approximately 1,217 versus 1,100, and mean `DesktopEvents` is approximately 1,118 versus 1,011.
- The association is weak rather than decisive: the strongest observed behavior correlation is `ActiveChannels` versus rating at approximately `-0.057`. Treat this as a high-use friction hypothesis, not proof that activity causes dissatisfaction.
- Ratings and response labels are consistent, so `Feedback_Rating` and `ResponseType` must not both be used as model predictors because they encode the same outcome.

### Where to investigate first

- **Japan tenant cohort:** NPS `-14.67` across 2,025 responses, with a 38.62% detractor rate. Validate locale, network quality, client mix, support experience, and release or incident exposure.
- **High Tech & Electronics:** NPS `+6.02` across 1,876 responses. This is a broad industry opportunity because it combines volume with a 10.10-point gap to the overall NPS.
- **IT Services and Software:** NPS `+8.97` and `+7.99`, respectively, with substantial response counts. Compare their client, country, and usage mix before assigning product ownership.
- **Passive population:** 9,032 responses, or 28.70% of the sample. This is the most visible conversion pool, but needs verbatims or follow-up reasons to identify the intervention.

### Limitations and next analysis

This dataset supports descriptive segmentation and behavioral hypothesis generation. It does not contain customer comments, feature satisfaction, incident data, latency, crash or call-quality telemetry, client version, or experiment assignments. Add those sources, classify detractor reasons, and then validate the strongest hypotheses using an out-of-time driver model or controlled intervention. Use response volume and confidence intervals before acting on any small segment.

### Action sequence

1. **Baseline the high-use cohorts.** Track desktop latency, failures, retries, crashes, sync issues, and support contacts for users with high reading, collaboration, and chat activity.
2. **Join customer evidence.** Link these cohorts to NPS comments, incident exposure, client version, locale, network quality, and feature-level satisfaction.
3. **Prioritize by reach and confidence.** Start with desktop, reading, collaboration, and chat workflows because their usage gaps repeat across all major customer segments.
4. **Test targeted improvements.** Use an experiment or matched pre/post design and measure NPS, detractor rate, task completion, latency, reliability, and support contacts.
5. **Protect segment outcomes.** Report results separately for Enterprise, SMC-SMB, SMC-Corporate, EDU, country, and user tenure so an overall gain does not hide harm in a subgroup.
